Q1. Write a Python function count_vowels that accepts a string and returns the number of vowels a, e, i, o, u, A, E, I, O, U in that string.

- Example: 'Welcome'
- Output: 3 vowels

```python
def count_vowels(string):
    string = string.lower()
    vowels = ['a','e','i','o','u']
    output = 0
    while len(vowels) != 0:
        if vowels[0] in string:
            output += string.count(vowels[0])
        vowels.pop(0)
    return f'Output: {output}'

count_vowels('Welcome')
```

Q6.

Write a Python function calculate_rectangle that accepts the length and width of a rectangle as arguments. Inside this function, create two nested functions:

- calculate_area: This function should calculate the area of the rectangle.
- calculate_perimeter: This function should calculate the perimeter of the rectangle.
- The main function should return both the area and the perimeter of the rectangle.

Expected Output:

Area: 15

Perimeter: 16


```python
def calculate_rectangle(a,b):
    def calculate_area():
        return a * b
    def calculate_perimeter():
        return (2 * a) + (2 * b)
    print(f'Area: {calculate_area()}')
    print(f'Perimeter: {calculate_perimeter()}')

calculate_rectangle(2,4)
```

Q8. Write a nested function to calculate the total hotel bill based on the number of nights.
A hotel charges Rs. 5,000 per night. The customer must also pay 10% tax on the room cost

Expected Output:
Room Cost: 15000
Tax: 1500.0
Total Bill: 16500.0

```python
def total_hotel_bill(nights):
    def room_cost():
        return nights * 5000
    def tax():
        return room_cost() * 0.1
    def total_bill():
        return room_cost() + tax()
    print("Room Cost: {}".format(room_cost()))
    print("Tax: {}".format(tax()))
    print("Total Bill: {}".format(total_bill()))

total_hotel_bill(3)
```

Q9.User Login Validation
A website checks a user's username and password before allowing access. Write a nested function to validate the login details

```python
database = {
    'john doe' : '123@jd'
}
def login_validation(username, password):
    def user_validation():
        if username.lower() in database:
            def password_validation():
                if password == database[username]:
                    return True
                else:
                    return False
            return password_validation()
        else:
            return False
    if user_validation() == False:
        return 'Either username or password is incorrect.'
    else:
        return 'Access Granted'
```

```python
login_validation('pizza','bermuda')
```

```python
login_validation('john doe','123@')
```

```python
login_validation('john doe','123@jd')
```

Q10.

Write a Python function shopping_bill(price, quantity, discount=0) that uses nested functions:

- An inner function calculate_total() → multiplies price * quantity.
- An inner function apply_discount(total) → subtracts discount from total (but never below 0).

The outer function should return the final bill amount.

Expected Output:

- Final Bill = 600 PKR (no discount)
- Final Bill = 500 PKR (with discount)

```python
def shopping_bill(price, quantity, discount=0):
    def calculate_total():
        return price * quantity
    def apply_discount(total):
        dis = total * (discount/100)
        final = total - dis
        return final
    print(f'Final Bill = {calculate_total()} PKR (no discount)')
    print(f'Final Bill = {apply_discount(calculate_total())} PKR (with discount)')
```

```python
shopping_bill(100,1,40)
```
