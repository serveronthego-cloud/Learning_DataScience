Q1 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]
check = 0
for i in users:
    if check < 3:
        print(f'Email sent to {i}')
    else:
        break
    check += 1
print("Limit Reached. Stop sending emails.")
```

Q2. Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] critical = 40

Ouput:

Temperature: 22

Temperature: 25

Temperature: 30

Critical temperature reached! Stop monitoring.

```python
temps = [22, 25, 30, 45, 28] ; critical = 40
for i in temps:
    if i >= 40:
        break
    else:
        print(f'Temperature: {i}')
print('Critical Temperature reached! Stop monitoring.')
```

Q3. Calculate total price of items in a shopping cart (Using For loop)

cart = [100, 250, 50, 300]

Output:

Total Price: 700

```python
cart = [100, 250, 50, 300]
sum_ = 0
for i in cart:
    sum_ = sum_ + i
print("Total Price:",sum_)
```

Q4. Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True}
]

Output:

Access granted to: Ali

Access granted to: John

```python
customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True} ]

for i in customers:
    if i["paid"] == False:
        continue
    else:
        print(f'Access granted to: {i['name']}')
```

Q5. Send personalized emails to a list of users (Using for loop)

users = ["Ali", "Sara", "John"]

Output:

Hello Ali, your subscription is active!
Hello Sara, your subscription is active!
Hello John, your subscription is active!

```python
users = ["Ali", "Sara", "John"]
for i in users:
    print(f'Hello {i}, your subscription is active!', end=" ")
```

Q6. Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

Output:

domains: ['gmail.com', 'yahoo.com', 'hotmail.com']



```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
domains = []
for i in emails:
    if '@' in i:
        ind = i.find('@')
        domains.append(i[ind+1:])
print(domains)
```

Q8. Create a multiplication table using a nested for loop.

numbers = [2, 3, 4]
multipliers = [1, 2, 3]

Expected Output:

2 x 1 = 2
2 x 2 = 4
2 x 3 = 6
3 x 1 = 3
3 x 2 = 6
3 x 3 = 9
4 x 1 = 4
4 x 2 = 8
4 x 3 = 12

Use a nested for loop.

```python
numbers = [2, 3, 4]; multipliers = [1, 2, 3]
for i in numbers:
    for j in multipliers:
        print(f'{i} x {j} = {i*j}')
```

Q9. Keep asking the user for a password until the correct password is entered.

correct_password = "python123"

Expected behaviour:

Enter password: abc
Incorrect password. Try again.

Enter password: 12345
Incorrect password. Try again.

Enter password: python123
Access granted!

Use a while loop.

```python
while True:
    passw = input("Enter password: ")
    if passw == 'python123':
        print("Access Granted!")
        break
    else:
        print("Try Again")
```

Q10. Count down from 5 to 1 and then display "Time's up!".

Expected Output:

5
4
3
2
1
Time's up!

Use a while loop.

```python
counter = 5
while counter > 0:
    print(counter)
    counter -= 1

print("Time's up")
```
